```python
from transformers import pipeline
import pandas as pd
```

Sentiment Analysis

```python
sentiment_classifier = pipeline("sentiment-analysis")

result1 = sentiment_classifier("Charleen is excited to learn more about Large Language Models")

result2 = sentiment_classifier("It is difficult to study AI Engineering!")

print(result1)
print(result2)
```

```python
#Used a pretrained sentiment analysis model. The pipeline automatically tokenizes the input text, passes it to the model, and converts the output logits into a human-readable label and confidence score
```

Zero-Shot Classification

```python
zeroshot_classifier = pipeline("zero-shot-classification", model = "facebook/bart-large-mnli")
```

```python
sequence_to_classify = "On December, Charleen will travel to see Almaty, Kazakhstan"
candidate_labels = ["travel", "cooking", "dancing"]
result = zeroshot_classifier(sequence_to_classify, candidate_labels)
print(result)
```

```python
#Zero-shot classification allows the classification of text into custom categories without training a new model. The model uses natural language inference to estimate which label best matches the input text.
```

Tokenization

```python
from transformers import AutoTokenizer

model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
sentence = "Charleen is excited to learn more about Large Language Models"

input_ids = tokenizer(sentence)
print(input_ids)
```

Tokens vs Token IDs

```python
tokens = tokenizer.tokenize(sentence)
print(tokens)
```

```python
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print(token_ids)
```

```python
decode_text = tokenizer.decode(token_ids)
print(decode_text)
```

```python
#Tokenization breaks text into smaller units called tokens. These tokens are then converted into numerical IDs from the model’s vocabulary. The model processes these IDs, not the raw text.
```

Special Tokens

```python
print(tokenizer.decode(101))
print(tokenizer.decode(102))
```

```python
#BERT adds special tokens such as CLS and SEP. CLS is commonly used for classification tasks, while SEP marks the end of a sentence or separates sentence pairs.
```

Manual Model Prediction

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
```

```python
#Load Tokenizer:
model_name = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
#Prepare Input:
sentence = "Charleen is excited to learn more about Large Language Models"

inputs = tokenizer(sentence, return_tensors="pt")

print(inputs)
```

```python
#Load Model
model = AutoModelForSequenceClassification.from_pretrained(model_name)
```

```python
#Run Prediction
with torch.no_grad():
    logits = model(**inputs).logits

predicted_class_ids = logits.argmax().item()

label = model.config.id2label[predicted_class_ids]
print(label)
```

```python
#nstead of using a pipeline, tokenizer and model was manually loaded. The text were tokenized, passed it into the model, got logits, selected the highest-scoring class using argmax, and mapped it back to the label using the model configuration.
```

Save and Reload the Model

```python
#Save tokenizer and model
model_directory = "my_saved_model"

tokenizer.save_pretrained(model_directory)
model.save_pretrained(model_directory)
```

```python
#Reload

my_tokenizer = AutoTokenizer.from_pretrained(model_directory)
my_model = AutoModelForSequenceClassification.from_pretrained(model_directory)
```

```python
#I saved both the tokenizer and model locally. To reload them correctly, I used .from_pretrained() because Hugging Face models require configuration and weight files to be restored properly.
```
